### Pre-Trained CNN architectures
- A pre-trained model is a model that has already been trained on a large dataset and learned useful patterns, weights, and features that can be reused on new, possibly different but related tasks.
- A **Pre-Trained CNN** Architecture is a convolutional neural network model that:

    - Has a fixed design/structure (architecture)—i.e., specific arrangement of layers (Conv, Pooling, FC).

    - Is trained on a large-scale dataset (like ImageNet) to learn general features.

    - Can be reused or fine-tuned for related downstream tasks

Read More about Pre-Trained CNN architectures  [here](https://vitalflux.com/different-types-of-cnn-architectures-explained-examples/)

```python
import numpy as np
import tensorflow
# from tensorflow import keras
from keras import Sequential
from keras_preprocessing import image
from keras.applications.resnet import ResNet50, preprocess_input, decode_predictions
```

```python
pre_trained_model = ResNet50(weights='imagenet', input_shape=(224, 224, 3))
```

```python
pre_trained_model.summary()  # Display the model architecture
```

```python
# Example usage of the pre-trained model
# Note: You would typically need to preprocess your input data before passing it to the model.

path = r'data/c.jpg'
img = image.load_img(path, target_size=(224, 224))
x = image.img_to_array(img) # Convert image to array so it can be processed by the model
x = np.expand_dims(x, axis=0) # Add batch dimension i.e, shape becomes (1, 224, 224, 3)
x = preprocess_input(x) # preprocess the input for ResNet50
```

```python
preds= pre_trained_model.predict(x) # Make predictions
print('Predicted:', decode_predictions(preds, top=3)[0]) # Decode and print the top 3 predictions
# Note: The decode_predictions function will give you the class labels and probabilities for the top predictions
```
