### 1. Simple Non-Sequential Model

```python
from keras.models import Model, Sequential
```

```python
from keras.layers import *
x = Input(shape=(3,))
```

```python
hidden1 = Dense(128, activation='relu')(x)
hidden2 = Dense(64, activation='relu')(hidden1)
```

```python
output1 = Dense(1, activation='linear')(hidden2)
output2 = Dense(1, activation='sigmoid')(hidden2)
```

```python
model = Model(inputs=x, outputs=[output1, output2])
```

```python
model.summary()
```

### 2. A complex functional Model
- branching 
- concatenation

#### Branch1

```python
input1 = Input(shape=(32, ))
dense1 = Dense(8, activation='relu')(input1)
dense2 = Dense(4, activation='relu')(dense1)
```

#### Branch2

```python
input2 = Input(shape=(128, ))
dense3 = Dense(64, activation='relu')(input2)
dense4 = Dense(32, activation='relu')(dense3)
dense5 = Dense(4, activation='relu')(dense4)
```

#### Concatenate : branch1 + branch2

```python
combined = concatenate([dense2, dense5])
```

```python
dense6 = Dense(2, activation='relu')(combined)
output = Dense(1, activation='sigmoid')(dense6)
```

```python
complex_model = Model(inputs=[input1, input2], outputs=[output])
```

```python
import netron
complex_model.save('model2.h5')
netron.start('model2.h5')
```
