```python
import numpy as np
import cv2
from PIL import Image
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

```python
# loading the image
img = mpimg.imread(r'img/p2.jpg')
plt.imshow(img)
img.shape
```

```python
type(img)
```

```python
# resizing the image
img = cv2.resize(img, (720, 524))
# displaying the resized image
plt.imshow(img)
img.shape
```

```python
cv2.imwrite('img/p2_resized.jpg', img) # saving the resized image
```

```python
# converting the image to grayscale
img = cv2.imread('img/p2_resized.jpg')
img_gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
# displaying the grayscale image
plt.imshow(img_gray, cmap='gray')
```
